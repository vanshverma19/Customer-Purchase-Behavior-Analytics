```python
# ==========================================
# 1. DATA QUALITY CHECK
# ==========================================

print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# ==========================================
# 2. DATA CLEANING
# ==========================================

# Remove duplicate rows
df = df.drop_duplicates()

# Convert Purchase_Date into datetime format
df["Purchase_Date"] = pd.to_datetime(
    df["Purchase_Date"],
    errors="coerce"
)

# Check invalid dates
print("Invalid dates:", df["Purchase_Date"].isnull().sum())

# Fill missing Age_Group values
df["Age_Group"] = df["Age_Group"].fillna("Unknown")

print("\nData cleaning completed.")
print("Final shape:", df.shape)
```

```python
# ==========================================
# 3. NUMERIC DATA VALIDATION
# ==========================================

numeric_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Discount_Percent",
    "Sales",
    "Profit",
    "Rating",
    "Profit_Margin",
    "Revenue_Before_Discount",
    "Discount_Amount"
]

df[numeric_columns].describe().T
```

```python
# ==========================================
# 4. FEATURE ENGINEERING
# ==========================================

# Net sales after discount
df["Net_Sales"] = df["Sales"]

# Revenue before discount
df["Gross_Revenue"] = df["Revenue_Before_Discount"]

# Profit per order
df["Profit_Per_Order"] = df["Profit"]

# Discount flag
df["Discount_Flag"] = np.where(
    df["Discount_Percent"] > 0,
    "Discounted",
    "No Discount"
)

# High-value order flag
median_sales = df["Sales"].median()

df["Order_Value_Type"] = np.where(
    df["Sales"] >= median_sales,
    "High Value",
    "Standard Value"
)

print("Feature engineering completed!")
```

```python
df[
    [
        "Order_ID",
        "Sales",
        "Profit",
        "Discount_Percent",
        "Net_Sales",
        "Discount_Flag",
        "Order_Value_Type"
    ]
].head(10)
```

```python
# ==========================================
# 5. BUSINESS KPIs
# ==========================================

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order_ID"].nunique()
total_customers = df["Customer_ID"].nunique()
total_units = df["Quantity"].sum()
average_order_value = df["Sales"].mean()
average_profit = df["Profit"].mean()
average_rating = df["Rating"].mean()
average_discount = df["Discount_Percent"].mean()

print("=" * 50)
print("          CUSTOMER PURCHASE ANALYTICS")
print("=" * 50)

print(f"Total Sales           : ₹{total_sales:,.2f}")
print(f"Total Profit          : ₹{total_profit:,.2f}")
print(f"Total Orders          : {total_orders:,}")
print(f"Total Customers       : {total_customers:,}")
print(f"Total Units Sold      : {total_units:,}")
print(f"Average Order Value   : ₹{average_order_value:,.2f}")
print(f"Average Profit/Order  : ₹{average_profit:,.2f}")
print(f"Average Rating        : {average_rating:.2f}")
print(f"Average Discount      : {average_discount:.2f}%")
```

```python
category_analysis = (
    df.groupby("Category")
      .agg(
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Profit", "sum"),
          Units_Sold=("Quantity", "sum"),
          Orders=("Order_ID", "nunique")
      )
      .sort_values("Total_Sales", ascending=False)
)

category_analysis
```

```python
plt.figure(figsize=(10, 6))

category_analysis["Total_Sales"].sort_values().plot(
    kind="barh"
)

plt.title("Sales by Product Category")
plt.xlabel("Sales")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

category_analysis["Total_Profit"].sort_values().plot(
    kind="barh"
)

plt.title("Profit by Product Category")
plt.xlabel("Profit")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

```python
region_analysis = (
    df.groupby("Region")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order_ID", "nunique"),
          Customers=("Customer_ID", "nunique")
      )
      .sort_values("Sales", ascending=False)
)

region_analysis
```

```python
plt.figure(figsize=(9, 5))

region_analysis["Sales"].plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
customer_type_analysis = (
    df.groupby("Customer_Type")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order_ID", "nunique"),
          Customers=("Customer_ID", "nunique"),
          Average_Order_Value=("Sales", "mean")
      )
      .sort_values("Sales", ascending=False)
)

customer_type_analysis
```

```python
subscription_analysis = (
    df.groupby("Subscription")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order_ID", "nunique"),
          Customers=("Customer_ID", "nunique"),
          Average_Order_Value=("Sales", "mean")
      )
)

subscription_analysis
```

```python
discount_analysis = (
    df.groupby("Discount_Flag")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order_ID", "nunique"),
          Average_Order_Value=("Sales", "mean"),
          Average_Profit=("Profit", "mean")
      )
)

discount_analysis
```

```python
plt.figure(figsize=(8, 5))

discount_analysis["Sales"].plot(kind="bar")

plt.title("Sales: Discounted vs Non-Discounted Orders")
plt.xlabel("Order Type")
plt.ylabel("Sales")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
monthly_sales = (
    df.groupby(["Year", "Month"])
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum")
      )
      .reset_index()
)

monthly_sales.head(20)
```

```python
monthly_trend = (
    df.groupby(
        df["Purchase_Date"].dt.to_period("M")
    )
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
)

monthly_trend["Purchase_Date"] = (
    monthly_trend["Purchase_Date"].astype(str)
)

monthly_trend.head()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_trend["Purchase_Date"],
    monthly_trend["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=60)

plt.tight_layout()
plt.show()
```

```python
top_products = (
    df.groupby("Product")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Quantity=("Quantity", "sum")
      )
      .sort_values("Sales", ascending=False)
      .head(10)
)

top_products
```

```python
plt.figure(figsize=(10, 6))

top_products["Sales"].sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("Product")

plt.tight_layout()
plt.show()
```

```python
customer_analysis = (
    df.groupby("Customer_ID")
      .agg(
          Customer_Name=("Customer_Name", "first"),
          Age=("Age", "first"),
          Gender=("Gender", "first"),
          City=("City", "first"),
          Region=("Region", "first"),
          Customer_Type=("Customer_Type", "first"),
          Subscription=("Subscription", "first"),
          Total_Sales=("Sales", "sum"),
          Total_Orders=("Order_ID", "nunique"),
          Total_Quantity=("Quantity", "sum"),
          Average_Order_Value=("Sales", "mean"),
          Total_Profit=("Profit", "sum"),
          Average_Rating=("Rating", "mean")
      )
      .reset_index()
)

customer_analysis.head()
```

```python
customer_analysis["Spending_Segment"] = pd.qcut(
    customer_analysis["Total_Sales"],
    q=4,
    labels=[
        "Low Value",
        "Medium Value",
        "High Value",
        "Premium"
    ]
)

customer_analysis.head()
```

```python
customer_analysis["Spending_Segment"].value_counts().sort_index()
```

```python
segment_analysis = (
    customer_analysis
    .groupby("Spending_Segment", observed=False)
    .agg(
        Customers=("Customer_ID", "count"),
        Total_Sales=("Total_Sales", "sum"),
        Average_Spend=("Total_Sales", "mean"),
        Average_Orders=("Total_Orders", "mean"),
        Total_Profit=("Total_Profit", "sum")
    )
)

segment_analysis
```
